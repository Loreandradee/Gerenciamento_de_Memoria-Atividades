```python
import gc  # Importamódulo de coleta de lixo
import sys  # Importa modulo pra pegar contagem de referências

# Parte 1: Criando a classe Objeto
class Objeto:
    def __init__(self, nome):
        self.nome = nome
        # Aqui acontece a alocação de um array grande pra simular uso de memória
        self.dados = [0] * 1000000  
        print(f"{self.nome} criado")  # Aqui faz a impressão de criação do objeto

    def __del__(self):
        # Aqui acontece a destruição do objeto
        print(f"{self.nome} destruído")  

        
        
# --- CENÁRIO 1: destruição automática quando a referência é removida ---
print("\n--- Cenário 1: destruição automática ---")
obj1 = Objeto("obj1")  # Criando objeto obj1
# Aqui faz mostrando a contagem de referências do obj1
print("Contagem de referências de obj1:", sys.getrefcount(obj1))  
# Aqui acontece a remoção da referência, esperando destruição automática
del obj1
print("obj1 deletado")  # Aqui faz mostrando que a referência foi deletada



# --- CENÁRIO 2: criando referências circulares ---
print("\n--- Cenário 2: referências circulares ---")
obj2 = Objeto("obj2")  # Criando objeto obj2
obj3 = Objeto("obj3")  # Criando objeto obj3



# Aqui acontece a criação de referência circular: obj2 aponta pra obj3 e obj3 aponta pra obj2
obj2.outro = obj3
obj3.outro = obj2



# Aqui faz mostrando contagem de referências antes de remover referências externas
print("Contagem de referências obj2:", sys.getrefcount(obj2))
print("Contagem de referências obj3:", sys.getrefcount(obj3))



# Aqui acontece remoção das referências externas, mas o ciclo ainda existe
del obj2
del obj3
print("obj2 e obj3 deletados, mas ciclo ainda está existindo")




# Aqui faz forçando a coleta de lixo pra limpar os objetos do ciclo
print("Forçando coleta de lixo...")
gc.collect()  # Chamando coletor geracional pra limpar ciclos



# --- CENÁRIO 3: mostrando contagem de referências vs coletor geracional ---
print("\n--- Cenário 3: contagem de referências vs coletor geracional ---")
obj4 = Objeto("obj4")  # Criando objeto obj4
obj5 = Objeto("obj5")  # Criando objeto obj5



# Aqui acontece a criação de referência de obj4 para obj5 (sem ciclo)
obj4.outro = obj5

# Aqui faz mostrando contagem de referências do obj5
print("Contagem de referências obj5:", sys.getrefcount(obj5))  # Conta referências



# Aqui acontece a remoção do obj5 diretamente (mas ainda referenciado por obj4)
del obj5
print("obj5 deletado (mas ainda referenciado por obj4)")



# Aqui faz removendo obj4 pra destruir obj5 também
del obj4
print("obj4 deletado, agora obj5 também será destruído")



# Aqui acontece mostrando estatísticas do coletor de lixo
print("\n--- Estatísticas do coletor de lixo ---")
stats = gc.get_stats()  # Pegando estatísticas de cada geração
for i, gen in enumerate(stats):
    print(f"Geração {i}: {gen}")
    
    

```
